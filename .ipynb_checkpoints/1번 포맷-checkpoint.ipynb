```python
import os, random
import numpy as np
import torch
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments
import evaluate
```

```python
# TODO: implement
def set_seed(seed: int=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

# TODO: implement
def build_dataset(tokenizer, train_size=500, eval_size=100, seed=42, max_lenght=256):
    def tokenize_fn(example):
        return tokenizer(
            example["text"],
            truncation=True,
            padding="max_length",
            max_length=max_length
        )
    ds = load_dataset("imdb")
    
    train_ds = ds["train"].shuffle(seed=seed).select(range(train_size))
    val_ds = ds["test"].shuffle(seed=seed).select(range(eval_size))

    train_tok = train_ds.map(tokenize_fn, batched=True)
    val_tok = val_ds.map(tokenize_fn, batched=True)

    train_tok = train_tok.rename_column("label", "labels")
    val_tok = val_tok.rename_column("label", "labels")

    train_tok.set_format(type="torch", columns=["input_ids", "attention_mask", "labels"])
    val_tok.set_format(type="torch", columns=["input_ids", "attention_mask", "labels"])

    return train_tok, val_tok
    
accuracy = evaluate.load("accuracy")
# TODO: implement
def compute_metrics(eval_preds):
    logits, labels = eval_preds
    preds = np.argmax(logits, axis=-1)

    return accuracy.compute(predictions=preds, references=labels)
    

# TODO: implement
def train_fn(model_name="distilbert-base-uncased", output_dir="./result", num_labels=2): 
    set_seed(42)
    device = torch.device("cpu")

    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=2)
    model.to(device)

    train, val = build_dataset(tokenizer)
    print(train)
    print(train[0])

    args = TrainingArguments(
        output_dir=output_dir,
        eval_strategy="epoch",
        num_train_epochs=1,
        per_device_train_batch_size=4,
        per_device_eval_batch_size=4,
        learning_rate=5e-5,
        save_strategy="no",
        logging_strategy="steps",
        logging_steps=10,
        # dataset_text_field="text"
    )

    trainer = Trainer(
        model=model,
        args=args,
        train_dataset=train,
        eval_dataset=val,
        processing_class=tokenizer
    )

    trainer.train()

    metrics = trainer.evaluate()
    print("Eval Metric: \n",smetrics)
    
    return metrics
    
```
